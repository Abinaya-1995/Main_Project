**Dataset Column/features Description:**

**1. Transaction and Product Details**

- **Invoice_ID**       : Unique ID for each transaction  

- **Invoice_Date**     : Date when the sale occurred  

- **City**             : Metro city where the transaction took place  

- **Store_Format**     : Type of retail store such as supermarket, hypermarket, or convenience store  

- **Category**         : Product category like food, beverages, or personal care  

- **Brand**            : Brand name of the product sold  

- **Channel**          : Sales channel used for purchase (Online or Offline)  

- **Payment_Mode**     : Payment method such as Cash, Card, or UPI  

- **Units**            : Number of units sold in the transaction  


**2.  Financial Metrics**

- **Cost_Price**       : Cost price per unit in INR  

- **Selling_Price**    : Selling price per unit in INR  

- **Revenue**          : Total revenue generated from the transaction  

- **Cost**             : Total cost incurred for the units sold  

- **Margin**           : Absolute profit earned on the transaction  

- **Margin_%**         : Profit margin expressed as a percentage  


**3. Operational and Inventory Metrics**

- **Stock_On_Hand**    : Inventory available at the time of sale  

- **Reorder_Level**    : Stock threshold that triggers replenishment  

- **Lead_Time_Days**   : Supplier lead time in days  


**4. Customer Details**

- **Customer_Age**     : Age of the customer  

- **Customer_Gender**  : Gender of the customer  

- **Loyalty_Flag**     : Indicates whether the customer is part of a loyalty program

#### 1. Handle missing values


```python
# Customer_Age column has 40081 null values

df['Customer_Age'] = df['Customer_Age'].fillna( round(df['Customer_Age'].median()))
```

```python
# Customer_Gender column has 5048 null values

df['Customer_Gender'] = df['Customer_Gender'].fillna(df['Customer_Gender'].mode()[0])
```

#### 2.Data type conversion

```python
# To convert Invoice_date (Object) data type as 'datetime' format

df['Invoice_Date'] = pd.to_datetime( df['Invoice_Date'] )
```

```python
# To convert Customer_Age (Float) data type as 'int' format

df['Customer_Age'] = df['Customer_Age'].astype(int)
```

#### 3. Split Invoice_Date Column (Date & Time) and Derive new fields

```python
# To split 'Time' from date column

df.insert(
    df.columns.get_loc('Invoice_Date') + 1,
    'Time',
    df['Invoice_Date'].dt.strftime('%I:%M %p')
)
```

```python
# To derive 'day name' from date column

df.insert(
    df.columns.get_loc('Invoice_Date') + 2,
    'Day_Name',
    df['Invoice_Date'].dt.day_name()
)
```

```python
# To derive 'month name' from date column

df.insert(
    df.columns.get_loc('Invoice_Date') + 3,
    'Month_Name',
    df['Invoice_Date'].dt.month_name()
)
```

```python
# To convert Invoice_date (datetime) as 'date' format

df['Invoice_Date'] = df['Invoice_Date'].dt.date
```

```python
df
```

#### 4.Trim all column names

```python
# To clean extra leading and trailing spaces

df.columns = df.columns.str.strip()
```

#### 5.Data Validation

```python
# Revenue Check

df['Calculated_Revenue'] = df['Units'] * df['Selling_Price']
```

```python
df['Revenue_Check'] = abs(df['Revenue'] - df['Calculated_Revenue']) < 0.05
```

```python
df['Revenue_Check'].value_counts()
```

```python
# Cost Check

df['Calculated_Cost'] = df['Units'] * df['Cost_Price']
```

```python
df['Cost_Check'] = abs(df['Cost'] - df['Calculated_Cost']) < 0.05
```

```python
df['Cost_Check'].value_counts()
```

```python
# Margin Check

df['Calculated_Margin'] = df['Revenue'] - df['Cost']
```

```python
df['Margin_Check'] = abs(df['Margin'] - df['Calculated_Margin']) < 0.05
```

```python
df['Margin_Check'].value_counts()
```

```python
# Margin_% Check

df['Calculated_Margin%'] = df['Margin'] / df['Revenue']
```

```python
df['Margin%_Check'] = abs(df['Margin_%'] - df['Calculated_Margin%']) < 0.0001
```

```python
df['Margin%_Check'].value_counts()
```

#### 6. Check the Skewness of Numeric Column

```python
# To check data spreadness whether positive skew or negative skew

df.skew(numeric_only=True)
```

#### 7. Check the outliers

```python
Q1 = df['Revenue'].quantile(0.25)
Q3 = df['Revenue'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5* IQR
upper = Q3 + 1.5* IQR

outliers_df = df[(df['Revenue'] < lower) | (df['Revenue'] > upper)]

outliers_df.shape[0]
```

### Bivariate Analysis → relation between two variables

```python
# revenue vs Category

revenue_by_category = (
    df.groupby('Category')['Revenue']
      .sum()
      .sort_values(ascending=False) / 1_000_000
)

plt.figure(figsize=(8,5), facecolor='#F8FFF8')

ax = sns.barplot(
    x=revenue_by_category.index,
    y=revenue_by_category.values,
    color='#81C784',
    edgecolor='#1B5E20',
    linewidth=1.5
)

labels = [f'{v:.2f}M' for v in revenue_by_category.values]

ax.bar_label(
    ax.containers[0],
    labels=labels

)

ax.set_facecolor('#FFFFFF')

plt.title(
    'Total Revenue by Category',
    fontsize=14,
    fontweight='bold',
    color='#1B5E20'
)

plt.xlabel('Category')
plt.ylabel('Revenue (Millions)')
plt.xticks(rotation=15)


plt.tight_layout()
plt.show()
```
